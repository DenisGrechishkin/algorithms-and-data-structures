## Задание 1: составить программу одного из методов сортировки в соответствии с вариантом. Выполнить топологическую сортировку на небольшом примере (8-10 элементов)

Вариант 7: написать программу для шейкерной сортировки

```python
def ShakerSorter(arr):
    start_time = time.time()
    startpoint = 0
    endpoint = len(arr) - 1
    while startpoint < endpoint:
        #проход справа налево
        for i in range(startpoint, endpoint):
            if arr[i] > arr[i + 1]:
                #меняем местами элементы массива, если предущий больше следующего
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
                
        #смещаем конечную точку массива на 1 влево
        endpoint -= 1
        
        #проход справа налево
        for i in range(endpoint - 1, startpoint - 1, -1):
            if arr[i] > arr[i + 1]:
                #меняем местами элементы массива, если предущий больше следующего
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
                
        #смещаем начальную точку массива на 1 вправо
        startpoint += 1
        
    end_time = time.time()
    print("Время сортировки:", end_time - start_time)
    return arr
```

воспользуемся функцией для сортировки массива

```python
list = [71, 22, 19, 37, 45, 99, 34, 11]
sorted = ShakerSorter(list)
sorted
```
