## Задание 1: Составить программу для исследования метода Крускала. 

```python
#создадим функцию определяющую минимальный остов графа, на вход которой подают массив из ребер
def KRASKAL(R, n_ver, n_reb):
    R_Sorted = sorted(R, key = lambda x: x[0])
    
    U = set()#список соединенных вершин
    D = {}#словарь списка изолированных групп вершин(для того чтобы понимать какая вершина к какой группе относится)
    T = []#cписок ребер остова
    
    #1 этап алгоритма Краскала - пары вершин, обе из которых - изолированные, 
    #или она из которых, принадлежит группе соединенных ребрами вершин, а другая - изолированная - соединяем ребром
    #таким образом мы избегаем образование циклов
    
    for r in R_Sorted:
        if r[1] not in U or r[2] not in U:#проверяем, является ли хотя-бы одна из вершин ребра изолированной
            if r[1] not in U and r[2] not in U:
                D[r[1]] = [r[1], r[2]]#формируем словарь 
                D[r[2]] = D[r[1]]#связываем номера вершин с одним списком
                
            else:#в противном случае добавляем в список вершину которая была изолирована
                if not D.get(r[1]):#1я изолирована
                    D[r[2]].append(r[1])#добавляем 1ю
                    D[r[1]] = D[r[2]]
                else:#2я изолирована
                    D[r[1]].append(r[2])#добавляем 2ю
                    D[r[2]] = D[r[1]]
                
            T.append(r)#ребро добавляем в остов
            U.add(r[1])#вершины добавляем в мн-во U
            U.add(r[2])
                
    #2 этап алгоритма - соединяем разные группы вершин ребрами минимального веса 
    for r in R_Sorted:
        if (r[1] in D[r[1]]) and (r[2] not in D[r[1]]):
            T.append(r)#добавляем ребро в остов
            gr1 = D[r[1]]
            D[r[1]] += D[r[2]]#объединяем списки вершин
            D[r[2]] += gr1
    if(len(T) == n_reb - n_ver + 1):
        print("Алгоритм работает верно")
        return T
    else:
        return("Алгоритм работает неверно")
```
