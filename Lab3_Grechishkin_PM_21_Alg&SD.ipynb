```python
def AvailableVersh(v, Graph):
    for i, weight in enumerate(Graph[v]):
        if(weight > 0):
            yield i

def FindMin(T, ProsmVersh):
    MIN = -1
    WAY_MIN = max(T)
    for i, t in enumerate(T):
        if t < WAY_MIN and i not in ProsmVersh:
            WAY_MIN = t
            MIN = i
    return MIN
```
